```c++
matrix train=readCSV("NNtrain.csv");
matrix test=readCSV("NNtest.csv");
// matDisplay(train);
// matDisplay(test);
matrix X_train,Y_train,X_test,Y_test;

split(train,X_train,Y_train);
split(test,X_test,Y_test);
matrix YT_real=Y_test;
// matDisplay(X_train);
// matDisplay(Y_test);
Zscale SclX;
Zscale SclY;

SclY.fit(Y_train);
SclY.scale(Y_train);
SclY.scale(Y_test);

SclX.fit(X_train);
SclX.scale(X_train);
SclX.scale(X_test);

// matDisplay(X_train);
// matDisplay(X_test);
// ReLU(X_train);
// matDisplay(X_train);
NeuralNetwork model;

model.fit(X_train,Y_train,0.001);
model.addLayer(25,"ReLU");
model.addLayer(20,"ReLU");
model.addLayer(10,"ReLU");
model.addLayer(1);
model.train(10000);

matrix pred=model.predict(X_test);
SclY.descale(pred);
matDisplay(pred);
// matDisplay(YT_real);
double err=RMSE(pred,YT_real);
cout<<"The Root Mean Squared Error is: "<<err<<endl;

```
