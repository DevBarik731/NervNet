## Regression Problem with NN

```c++
matrix train=readCSV("data/NNtrain.csv");
matrix test=readCSV("data/NNtest.csv");
// matDisplay(train);
// matDisplay(test);
matrix X_train,Y_train,X_test,Y_test;

split(train,X_train,Y_train);
split(test,X_test,Y_test);
matrix YT_real=Y_test;
// matDisplay(X_train);
// matDisplay(Y_test);
Zscale SclX;
Zscale SclY;

SclY.fit(Y_train);
SclY.scale(Y_train);
SclY.scale(Y_test);

SclX.fit(X_train);
SclX.scale(X_train);
SclX.scale(X_test);

// matDisplay(X_train);
// matDisplay(X_test);
// ReLU(X_train);
// matDisplay(X_train);
NeuralNetwork model;

model.fit(X_train,Y_train,0.001);
model.addLayer(25,"ReLU");
model.addLayer(20,"ReLU");
model.addLayer(10,"ReLU");
model.addLayer(1);
model.train(500);

matrix pred=model.predict(X_test);
SclY.descale(pred);
matDisplay(pred);
// matDisplay(YT_real);
double err=RMSE(pred,YT_real);
cout<<"The Root Mean Squared Error is: "<<err<<endl;

```

## Classification Problem

```c++
matrix train=readCSV("data/IEtrain.csv");
matrix test=readCSV("data/IEtest.csv");
// matDisplay(train);
// matDisplay(test);
matrix X_train,Y_train,X_test,Y_test;

split(train,X_train,Y_train);
split(test,X_test,Y_test);
matrix YT_real=Y_test;
// matDisplay(X_train);
// matDisplay(Y_test);
Zscale SclX;

SclX.fit(X_train);
SclX.scale(X_train);
SclX.scale(X_test);

// matDisplay(X_train);
// matDisplay(X_test);
// ReLU(X_train);
// matDisplay(X_train);
NeuralNetwork model;

model.fit(X_train,Y_train,0.001);
model.addLayer(25,"ReLU");
model.addLayer(20,"ReLU");
model.addLayer(20,"ReLU");
model.addLayer(10,"ReLU");
model.addLayer(1,"sigmoid");
model.train(3000);

matrix pred=model.predict(X_test);

sigToBin(pred);
matDisplay(pred);

double acc=accuracy(pred,Y_test);
cout<<"Accuracy of model is "<<(acc*100)<<"%"<<endl;

// matDisplay(YT_real);
```

```c++
matrix tst=readCSV("data/tst.csv");
SclX.scale(tst);
matrix Y=model.predict(tst);
sigToBin(Y);
matDisplay(Y);
```
