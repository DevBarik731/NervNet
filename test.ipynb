## Regression Problem with NN

```c++
matrix train=readCSV("data/NNtrain.csv");
matrix test=readCSV("data/NNtest.csv");

matrix X_train,Y_train,X_test,Y_test;

split(train,X_train,Y_train);
split(test,X_test,Y_test);
matrix Y_test_real=Y_test;
matrix Y_train_real=Y_train;

Zscale SclX;
Zscale SclY;

SclY.fit(Y_train);
SclY.scale(Y_train);
SclY.scale(Y_test);

SclX.fit(X_train);
SclX.scale(X_train);
SclX.scale(X_test);


NeuralNetwork model;

model.fit(X_train,Y_train,0.001,128);
model.addLayer(25,"ReLU");
model.addLayer(20,"ReLU");
model.addLayer(1);
model.train(100);

matrix pred_test=model.predict(X_test);
SclY.descale(pred_test);
matrix pred_train=model.predict(X_train);
SclY.descale(pred_train);

double er_test=RMSE(pred_test,Y_test_real);
double er_train=RMSE(pred_train,Y_train_real);
cout<<"The Root Mean Squared Error in test set is: "<<(er_test)<<endl;
cout<<"The Root Mean Squared Error in train set is: "<<(er_train)<<endl;
```

## Classification Problem

```c++
matrix train=readCSV("data/IEtrain.csv");
matrix test=readCSV("data/IEtest.csv");

matrix X_train,Y_train,X_test,Y_test;

split(train,X_train,Y_train);
split(test,X_test,Y_test);
matrix YT_real=Y_test;

Zscale SclX;

SclX.fit(X_train);
SclX.scale(X_train);
SclX.scale(X_test);

NeuralNetwork model;

model.fit(X_train,Y_train,0.001,512);
model.addLayer(25,"ReLU");
model.addLayer(20,"ReLU");
model.addLayer(20,"ReLU");
model.addLayer(10,"ReLU");
model.addLayer(1,"sigmoid");
model.train(50);

matrix pred_test=model.predict(X_test);
matrix pred_train=model.predict(X_train);
pred_test.sigToBin();
pred_train.sigToBin();


double acc_test=accuracy(pred_test,Y_test);
double acc_train=accuracy(pred_train,Y_train);
cout<<"Accuracy of model on test set is : "<<(acc_test*100)<<"%"<<endl;
cout<<"Accuracy of model on train set is : "<<(acc_train*100)<<"%"<<endl;


```
